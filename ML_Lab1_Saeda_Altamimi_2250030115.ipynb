```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

# Ensure both lists have the same length
if len(study_hours) != len(marks):
    raise ValueError("Error: study_hours and marks lists must have the same length.")

# 1. Number of samples
num_samples = len(study_hours)
print(f"Number of samples: {num_samples}")

# 2. Average study hours
avg_hours = sum(study_hours) / num_samples if num_samples > 0 else 0
print(f"Average study hours: {avg_hours:.2f}")

# 3. Average marks
avg_marks = sum(marks) / num_samples if num_samples > 0 else 0
print(f"Average marks: {avg_marks:.2f}")

# 4. Highest and lowest marks
highest_mark = max(marks) if marks else None
lowest_mark = min(marks) if marks else None
print(f"Highest mark: {highest_mark}")
print(f"Lowest mark: {lowest_mark}")

# 5 & 6. Display each sample with Pass/Fail
print("\nResults:")
for i, (h, m) in enumerate(zip(study_hours, marks), start=1):
    status = "Pass" if m >= 60 else "Fail"
    print(f"Student {i}: Study Hours = {h}, Marks = {m}, Status = {status}")

# 7. Count the number of passes
pass_count = sum(1 for m in marks if m >= 60)
print(f"\nNumber of students with marks >= 60: {pass_count}")
```
