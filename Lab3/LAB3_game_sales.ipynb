```python
plt.figure(figsize=(8,5))
sns.histplot(df['Year_of_Release'],bins=20)
plt.title("Distribution of games released per year")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df['Genre'], bins=20)
plt.title("Game Genre")
plt.title("Game Genre Distribution", fontsize=18, fontweight='bold')
plt.xlabel("Genre", fontsize=14)
plt.ylabel("Count", fontsize=14)
plt.show()
```

```python
# User/record count by sales region
Regions_users = {
    'North America': (df['NA_Sales'] > 0).sum(),
    'Europe': (df['EU_Sales'] > 0).sum(),
    'Japan': (df['JP_Sales'] > 0).sum(),
    'Other': (df['Other_Sales'] > 0).sum()
}

# Plot user count by region
plt.figure(figsize=(10,5))

plt.bar(Regions_users.keys(), Regions_users.values())

plt.title("Video Game Consumers by Region")
plt.xlabel("Region")
plt.ylabel("User_Count")

plt.show()
```

```python
Platform_User_Count = df.groupby('Platform')['User_Count'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
Platform_User_Count.plot(kind='bar')
plt.title("Users Per Platform")
plt.ylabel("Revenue")
plt.show()

Platform_User_Count
```

```python
Publisher_Sales = df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False)

Publisher_Sales.head(10)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Critic_Score', y='Global_Sales', data=df)
plt.title("Critic Score vs Global Sales")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['JP_Sales', 'Global_Sales']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
year_critic_score = df.groupby('Year_of_Release')['Critic_Score'].sum()

plt.figure(figsize=(10,5))

year_critic_score.plot(kind='line')

plt.title("Average Critic Score by Year")
plt.xlabel("Year_of_Release")
plt.ylabel("Average Critic Score")

plt.show()
```
